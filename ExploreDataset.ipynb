## BraTS_2019

### HGG = High-Grade Glioma
These are aggressive, fast-growing brain tumors.
Characteristics:
Grow rapidly
Highly invasive
Usually enhance strongly on T1ce
Large regions of necrosis (dead tissue)
More edema (swelling)
Poor prognosis
The most common HGG is:
Glioblastoma (GBM) — WHO Grade IV (under the older WHO classification)

Large tumor with irregular shape.
### LGG = Low-Grade Glioma
These are much slower-growing tumors.
Characteristics:
Slower growth
Less aggressive
Often little or no enhancement after contrast
Less necrosis
Better prognosis
Older WHO grades:
Grade II
Grade III (depending on the classification era)
